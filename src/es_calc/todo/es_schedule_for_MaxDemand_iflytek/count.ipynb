```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("./data/schedule_result_scale_10000.csv", index_col=0, parse_dates=[0])
```

```python
# df["energy"] = df["power_opt"] /12
df['charge'] = np.where(df['power_opt'] < 0, df['power_opt'], 0)  # 小于0的值保留，否则填0
df['discharge'] = np.where(df['power_opt'] > 0, df['power_opt'], 0)  # 大于0的值保留，否则填0
```

```python
monthly_cost = df['charge'].resample('M').sum()

# 提取年份和月份，并打印月份数字和对应的 cost
for date, cost in monthly_cost.items():
    year = date.year
    month = date.month
    print(f"Year: {year}, Month: {month}, Total Cost: {cost/12}")
```
