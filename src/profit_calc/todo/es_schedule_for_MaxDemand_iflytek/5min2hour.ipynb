```python
import pandas as pd
```

```python
# 1. 读取 CSV 文件
df = pd.read_csv('demand_load.csv')

# 2. 将 time 列转换为 datetime 类型（如果还不是）
df['time'] = pd.to_datetime(df['time'])

# 3. 将 time 列设为索引（resample 需要 DatetimeIndex）
df.set_index('time', inplace=True)

# 4. 按小时重采样，并对 value 列进行聚合（例如求和、平均等）
# 常见做法：如果是负荷（load）数据，通常用平均值或最大值；如果是累计量（如电量），则用求和。
# 这里以平均值为例：
hourly_df = df['value'].resample('H').mean()  # 'H' 表示 hourly

# 如果你希望保留为 DataFrame 而不是 Series，并且把时间变回普通列：
hourly_df = hourly_df.reset_index()
```

```python
# 5. 保存结果（可选）
hourly_df.to_csv('demand_load_1.csv', index=False)

# 打印前几行查看结果
print(hourly_df.head())
```
