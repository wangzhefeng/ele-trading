```python
import pandas as pd
```

```python
exp_name = "estimate1122"
node_name = "route_10"
es_scale = 625
```

```python
df = pd.read_csv(f"./data/{exp_name}/{node_name}/opt_result/es_scale_experiment/schedule_result_scale_{es_scale}.csv")
df.rename(columns={"power_opt": "value"}, inplace=True)
df['time'] = pd.to_datetime(df['time'])
df.set_index('time', inplace=True)
```

```python
result = df.groupby([df.index.year, df.index.month, df.index.hour])['value'].mean()

# 转成更清晰的 DataFrame（可选）
result = result.rename_axis(['year', 'month', 'hour']).reset_index()
```

```python
result.to_csv("result_625.csv")
```
